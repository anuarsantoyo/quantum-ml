# 19c — σ-channel calibration: does matching the real fit-error scale recover μ?

**Series 19 — autonomous improvement round (goal: honest uncertainties on the real dataset).**
Trial c: 19b proved the real-data μ attractor (NLL min at μ/μ_true ≈ 0.30) lives in the
σ_fit channel — real fit-errors are 1.04–4.25× the simulator's at μ_true (gap grows with T),
and the count channel is dead. This trial scales the simulator's σ_fit by c_exp (from the
19b autopsy) and re-runs the μ-score sweep.

**Key insight:** σ-scale calibration (c=c_exp) does NOT move the μ-likelihood mode — even with the
simulator's σ matched to real fit-errors at μ_true (verified: sim_σ_med ≈ real_med in all 6 exps),
NLL still minimizes at μ/μ_true = 0.5 everywhere. The attractor is the SHAPE of σ(μ), not its scale.
C1 (calibrated count score) pulls μ to 0.54–0.69× honest-wide and fixes 1nW T100 γ (8.49 vs 8.5);
C2 (calibrated σ-gradient) is a true restoring force at low T (T20: 1.04×/0.12σ, 1.13×/0.40σ) but
destabilizes at high T (T60/T100 collapse to the floor, |Δμ|/σ up to 323) — the pull ∝ c·σ/n is
un-damped. Verdict: NEITHER branch → μ unidentifiable under any Lorentzian-σ model → 19d = honest-wide
D1 final answer on all 14 exps + failure mode #8.


## 1. CURRENT STATE ANALYSIS (written before this trial)

### 1.1 What 19b established (real data, 6 exps, 3 μ-score designs)
- **The μ attractor is real**: NLL(μ) at fixed γ_true peaks at μ/μ_true = 0.30 in ALL six
  experiments (NLL 2.6–3.3 — healthy fit, not flatness). The 18-series "½ + huge σ_μ"
  was the optimizer caught between this low-μ mode and a useless count channel.
- **Count channel is dead**: grad_count ≈ 0 everywhere. With 1100–2500 real data points
  the KDE responsibility B = w.mean(0) ≈ uniform, so the REINFORCE count score
  (n−μ)/σ_ref² averages to zero. μ information can ONLY flow through the σ_fit channel.
- **The σ_fit channel is mismatched**: sim σ_fit median @ μ_true vs real fit-error median:
  1.04 / 2.05 / 2.36 (1nW T20/T60/T100), 2.15 / 3.44 / 4.25 (3nW T20/T60/T100). The gap
  GROWS with T. The Lorentzian model can only match the real error scale by photon
  starvation (lower μ) → the likelihood mode sits at ~0.3 μ_true.
- **Score designs**: D0 (σ_ref=10) tight-wrong μ (|Δμ|/σ_μ 16.6 @ 3nW T100); D1
  (σ_ref=σ_prop) honest-but-useless (σ_μ ≈ 0.5 μ_true, μ̂ still ~0.45); D2 (raw σ
  gradient) collapses μ to the floor (|Δμ|/σ_μ up to 557) — the raw gradient descends
  INTO the low-μ mode. All consistent with ONE mechanism: σ-channel mismatch.

### 1.2 The fix this trial tests
Scale the simulator's σ_fit by c_exp = real_err_med / sim_σ_med@μ_true (measured in the
19b autopsy, per experiment). Then at μ_true, c·σ_sim ≈ real:
- C1 (calibration only, D1 count score): does the likelihood mode move to μ_true?
  (The count score is dead, so μ may not MOVE — but the Fisher should show the
  information content of the corrected σ channel.)
- C2 (calibration + D2 σ-gradient): at μ_true, d_s = real − c·σ_sim ≈ 0 → zero pull;
  BELOW μ_true, c·σ_sim > real → d_s < 0 → dσ'/dμ < 0 → gradient > 0 → μ rises;
  ABOVE, the reverse. The gradient becomes a RESTORING force with a stable fixed point
  at μ_true. Prediction: μ̂ → μ_true with honest coverage.
- Caveat (noted honestly): c_exp is calibrated on the same 6 exps being tested. The
  generalization test is 19d on all 14 exps — c(T) grows smoothly with T, so if it is a
  real law, the 8 held-out exps will confirm it.

### 1.3 γ side (unchanged, for the record)
3nW γ̂ 11.5–14.6 vs 14.1 with |Δγ|/σ ≤ 0.79 in D1 (good); 1nW still low (4.0–8.0 vs 8.5)
— the known failure-mode #6, untouched by this trial.


## 2. PROPOSAL — σ-channel calibration sweep (design, predictions, decision rule)

### 2.1 Design
- **Targets**: REAL scans, the same 6 experiments as 19b (1nW/3nW T20/T60/T100).
- **Simulator change (data side only)**: sim σ_fit is multiplied by c before entering the
  KDE and the score channels. c = 1 for the control, c = c_exp (per-exp, from 19b
  autopsy) for the calibrated arms. 18c optimizer otherwise EXACTLY as 19b.
- **Designs**:
  - **C0**: D1 μ score (σ_ref=σ_prop), c=1 → honest-wide control (reproduces 19b-D1).
  - **C1**: D1 μ score, c=c_exp → calibration alone.
  - **C2**: D2 μ score (count + σ-gradient, dσ'/dμ = −c·σ/2n), c=c_exp → calibrated
    restoring gradient.
- **Post-calibration check** (cheap, before the sweep): NLL at μ/μ_true ∈ {0.5, 1.0, 1.5}
  with γ=γ_true and c=c_exp — does the likelihood mode move from 0.30 to ~1.0?
- **Uncertainty**: Fisher at the fitted point with the design's own μ score and scaled σ
  (honest coverage), raw γ-score, spread diagnostic — as 19b.

### 2.2 Predictions
- **Check**: NLL(μ) with c=c_exp is minimized at μ/μ_true ≈ 1.0 (mode moved).
- **C0**: reproduces 19b-D1 (μ̂ ≈ 0.45–0.52, σ_μ ≈ 0.5 μ_true, honest but useless).
- **C1**: likelihood mode at ~1.0, but the dead count score leaves μ̂ near init → wide
  σ_μ; Fisher tells us how much information the corrected σ channel really carries.
- **C2**: μ̂ converges to μ_true; σ_μ honest (|Δμ|/σ_μ ≲ 1.5, σ_μ ≲ 0.3 μ_true).

### 2.3 Decision rule → 19d
- **C2 recovers** (|Δμ|/σ_μ ≲ 1.5 AND σ_μ ≲ 0.3·μ_true in ≥5/6 exps) → adopt the
  σ-calibrated simulator + σ-gradient μ score; 19d validates on ALL 14 exps (8 held-out
  test the c(T) law) with full Fisher.
- **C1 only** (mode at 1.0 but μ still not moving) → 19d adds the σ-gradient to C1's
  score and/or tunes the c(T) law, still closed loop.
- **Neither** → μ is unidentifiable under any Lorentzian-σ model; 19d reports the D1
  honest-wide answer as final and documents the σ-channel mismatch as failure mode #8.

```python
# ============================================================
# 4. REAL TARGETS (15-series load & filter) + machinery import
# ============================================================
def load_real_targets(exp):
    d = np.genfromtxt(exp['data_file'])
    fwhm_mhz = d[:, 0] * 1000.0
    err_mhz = d[:, 1] * 1000.0
    ok = ~np.isnan(fwhm_mhz) & ~np.isnan(err_mhz) & (fwhm_mhz > 0)
    filt = ok & (err_mhz / fwhm_mhz < 10.0)
    target_f = torch.tensor(fwhm_mhz[filt], dtype=torch.float32)
    target_s = torch.tensor(err_mhz[filt], dtype=torch.float32)
    if TARGET_CAP is not None and len(target_f) > TARGET_CAP:
        idx = np.random.default_rng(12345).choice(len(target_f), TARGET_CAP, replace=False)
        target_f = target_f[idx]
        target_s = target_s[idx]
    return target_f, target_s

def target_bandwidths(target_f, target_s):
    n = len(target_f)
    SCOTT = n ** (-1.0 / 6.0)
    H_F = float(target_f.std()) * SCOTT
    H_S = max(float(target_s.std()) * SCOTT, H_S_MIN)
    return H_F, H_S

def qs(x):
    return [float(np.quantile(x.detach().numpy() if torch.is_tensor(x) else x, q)) for q in (0.1, 0.25, 0.5, 0.75, 0.9)]

print('real-target loader ready')
```

```python
# ============================================================
# 5. LIKELIHOOD + μ-SCORE DESIGNS (with σ-scale c) + POST-CALIBRATION CHECK
# ============================================================
def kde_scores(sim_f, sim_s, sim_n, sim_df, sim_ds, data_f, data_s, h_f, h_s, mu, sigma_prop, gamma_scale=True):
    d_f = data_f[:, None] - sim_f[None, :]
    d_s = data_s[:, None] - sim_s[None, :]
    W = torch.exp(-0.5 * (d_f / h_f) ** 2 - 0.5 * (d_s / h_s) ** 2)
    w = W / W.sum(dim=1, keepdim=True).clamp_min(1e-12)
    score = (sim_n[None, :] - mu) / sigma_prop ** 2
    s_mu = (w * score).sum(dim=1)
    if gamma_scale:
        dlogG = ((d_f * sim_df[None, :]) / h_f + (d_s * sim_ds[None, :]) / h_s) / H_REF
    else:
        dlogG = (d_f * sim_df[None, :]) / h_f ** 2 + (d_s * sim_ds[None, :]) / h_s ** 2
    s_gamma = (w * dlogG).sum(dim=1)
    logp = torch.log((W.sum(dim=1) / len(sim_f)).clamp_min(1e-30))
    return s_mu, s_gamma, -logp.mean(), w

def grad_mu_design(score_variant, c, w, sim_n, sim_s, data_s, h_s, mu, sigma_prop):
    '''Scalar μ gradient. sim_s is ALREADY scaled by c.'''
    B = w.mean(dim=0)
    if score_variant == 'D1':
        score = (sim_n - mu) / sigma_prop ** 2
        return float(max(min(-(B - B.mean()) @ score, CLIP), -CLIP))
    if score_variant == 'D2':
        score = (sim_n - mu) / SIGMA_REF ** 2
        grad_count = -(B - B.mean()) @ score
        d_s = data_s[:, None] - sim_s[None, :]
        dsigma_dmu = -sim_s / (2.0 * sim_n.clamp_min(1.0))   # d(c·σ)/dμ = −c·σ/2n
        s_sigma = (w * (d_s / h_s) * dsigma_dmu[None, :] / H_REF).sum(dim=1).mean()
        return float(max(min(-(grad_count + s_sigma), CLIP), -CLIP))
    raise ValueError(score_variant)

def mu_score_point(score_variant, c, w, sim_n, sim_s, data_s, h_s, mu, sigma_prop):
    '''Per-data-point μ score (Fisher). sim_s is ALREADY scaled by c.'''
    if score_variant == 'D1':
        score = (sim_n[None, :] - mu) / sigma_prop ** 2
        return (w * score).sum(dim=1)
    if score_variant == 'D2':
        score = (sim_n[None, :] - mu) / SIGMA_REF ** 2
        s_count = (w * score).sum(dim=1)
        d_s = data_s[:, None] - sim_s[None, :]
        dsigma_dmu = -sim_s[None, :] / (2.0 * sim_n[None, :].clamp_min(1.0))
        s_sigma = (w * (d_s / h_s) * dsigma_dmu / H_REF).sum(dim=1)
        return s_count + s_sigma
    raise ValueError(score_variant)

def calibration_check(exp, target_f, target_s, pool):
    '''NLL at mu/mu_true in {0.5, 1.0, 1.5} with c=c_exp, gamma=gamma_true.'''
    H_F, H_S = target_bandwidths(target_f, target_s)
    mu_true, sigma_prop, lam, gamma_true = exp['mu_true'], exp['sigma_prop'], exp['lam'], exp['gamma_true']
    c = C_EXP[exp['name']]
    rows = []
    rng = np.random.default_rng(4321)
    for ratio in (0.5, 1.0, 1.5):
        mu = ratio * mu_true
        tasks, ns = [], []
        for _ in range(N_AUTO):
            u, b, n = draw_fixed_noise(mu, sigma_prop, lam, rng)
            tasks.append((gamma_true, u.numpy(), b.numpy()))
            ns.append(n)
        res = parallel_map(pool, tasks)
        ft = torch.tensor([r[0] for r in res], dtype=torch.float32)
        si_t = torch.tensor([r[1] for r in res], dtype=torch.float32) * c
        nt = torch.tensor(ns, dtype=torch.float32)
        dg_t = torch.tensor([r[2] for r in res], dtype=torch.float32)
        ds_t = torch.tensor([r[3] for r in res], dtype=torch.float32)
        _, _, nll_val, _ = kde_scores(ft, si_t, nt, dg_t, ds_t, target_f, target_s,
                                      H_F, H_S, mu, sigma_prop, gamma_scale=True)
        rows.append(dict(mu_ratio=ratio, nll=float(nll_val), sim_s_med=float(torch.median(si_t))))
    return dict(exp=exp['name'], c=c, target_s_med=float(torch.median(target_s)), rows=rows)

print('score designs + calibration check ready')
```

```python
# ============================================================
# 6. ONE EXPERIMENT: REAL targets, σ-scaled simulator, μ score per design
# ============================================================
def run_experiment(exp, design, target_f, target_s, pool):
    score_variant, c = DESIGN_SPEC[design]
    if c is None:
        c = C_EXP[exp['name']]
    mu_true, sigma_prop = exp['mu_true'], exp['sigma_prop']
    lam, gamma_true = exp['lam'], exp['gamma_true']
    mu_init, gamma_init = 0.5 * mu_true, 0.5 * gamma_true
    H_F, H_S = target_bandwidths(target_f, target_s)

    mu_val, gamma_val = float(mu_init), float(gamma_init)
    history = []
    t0 = time.time()
    for step in range(N_ITER):
        rng2 = np.random.default_rng(SEED + step)
        tasks, ns = [], []
        for _ in range(N_RUNS):
            u, b, n = draw_fixed_noise(mu_val, sigma_prop, lam, rng2)
            tasks.append((gamma_val, u.numpy(), b.numpy()))
            ns.append(n)
        res = parallel_map(pool, tasks)
        ft = torch.tensor([r[0] for r in res], dtype=torch.float32)
        si_t = torch.tensor([r[1] for r in res], dtype=torch.float32) * c   # σ-scale
        nt = torch.tensor(ns, dtype=torch.float32)
        dg_t = torch.tensor([r[2] for r in res], dtype=torch.float32)
        ds_t = torch.tensor([r[3] for r in res], dtype=torch.float32)

        s_mu, s_gamma, nll_val, w = kde_scores(ft, si_t, nt, dg_t, ds_t, target_f, target_s,
                                               H_F, H_S, mu_val, sigma_prop, gamma_scale=True)
        grad_mu = grad_mu_design(score_variant, c, w, nt, si_t, target_s, H_S, mu_val, sigma_prop)
        grad_gamma = float(max(min(-s_gamma.mean(), CLIP), -CLIP))

        lr_mu_decay = LR_MU * (1.0 - step / N_ITER)
        mu_val -= lr_mu_decay * grad_mu
        mu_val = max(1.0, min(200.0, mu_val))
        gamma_val -= LR_GAMMA * (1.0 - GAMMA_ANNEAL * step / N_ITER) * grad_gamma
        gamma_val = max(0.1, min(100.0, gamma_val))
        history.append(dict(step=step, mu=mu_val, gamma=gamma_val, nll=float(nll_val)))

    return dict(exp=exp['name'], power=exp['power'], design=design, c=c,
                score_variant=score_variant,
                mu_true=mu_true, gamma_true=gamma_true, sigma_prop=sigma_prop, lam=lam,
                mu_init=mu_init, gamma_init=gamma_init, mu_final=mu_val, gamma_final=gamma_val,
                nll_final=float(nll_val), t_elapsed=time.time() - t0,
                target_f=target_f, target_s=target_s, H_F=H_F, H_S=H_S, history=history)

print('run_experiment ready')
```

```python
# ============================================================
# 7. FISHER / CRB at the fitted point (design's own μ score, scaled σ;
#    raw γ-score for 15-series comparability) + spread diagnostic
# ============================================================
def fisher_at_fitted(r, pool):
    mu_f, gamma_f = r['mu_final'], r['gamma_final']
    sp, lam = r['sigma_prop'], r['lam']
    score_variant, c = r['score_variant'], r['c']
    rng = np.random.default_rng(FISHER_BASE_SEED)
    tasks, ns = [], []
    for _ in range(M_FINAL):
        u, b, n = draw_fixed_noise(mu_f, sp, lam, rng)
        tasks.append((gamma_f, u.numpy(), b.numpy()))
        ns.append(n)
    res = parallel_map(pool, tasks)
    ft = torch.tensor([x[0] for x in res], dtype=torch.float32)
    si_t = torch.tensor([x[1] for x in res], dtype=torch.float32) * c
    nt = torch.tensor(ns, dtype=torch.float32)
    dg_t = torch.tensor([x[2] for x in res], dtype=torch.float32)
    ds_t = torch.tensor([x[3] for x in res], dtype=torch.float32)
    target_f, target_s = r['target_f'], r['target_s']

    _, s_gamma, _, w = kde_scores(ft, si_t, nt, dg_t, ds_t, target_f, target_s,
                                  r['H_F'], r['H_S'], mu_f, sp, gamma_scale=False)
    s_mu = mu_score_point(score_variant, c, w, nt, si_t, target_s, r['H_S'], mu_f, sp)
    s = torch.stack([s_mu, s_gamma], dim=1)
    J = s.T @ s / len(target_f)
    Jinv = torch.linalg.inv(J + 1e-8 * torch.eye(2))
    std_mu = float(math.sqrt(Jinv[0, 0]))
    std_gamma = float(math.sqrt(Jinv[1, 1]))
    corr = float(Jinv[0, 1] / torch.sqrt(Jinv[0, 0] * Jinv[1, 1]))
    spread_ratio = float(ft.std()) / float(target_f.std())
    return dict(exp=r['exp'], design=r['design'], c=c, mu_f=mu_f, gamma_f=gamma_f,
                mu_true=r['mu_true'], gamma_true=r['gamma_true'], n_target=len(target_f),
                std_mu=std_mu, std_gamma=std_gamma, corr=corr, spread_ratio=spread_ratio,
                dmu_sigma=abs(mu_f - r['mu_true']) / std_mu,
                dgamma_sigma=abs(gamma_f - r['gamma_true']) / std_gamma)

print('fisher_at_fitted ready')
```

```python
# ============================================================
# 8. RUN: calibration check, then 3 designs x 6 experiments, then Fisher
# ============================================================
import multiprocessing as _mp
from concurrent.futures import ProcessPoolExecutor as _PPE

N_CPUS = os.cpu_count() or 4
N_WORKERS = 4

t_all = time.time()

# ---- 8a. POST-CALIBRATION CHECK (does the likelihood mode move to ~1.0?) ----
print('CALIBRATION CHECK ...', flush=True)
checks = []
with _PPE(max_workers=N_WORKERS, mp_context=_mp.get_context('fork'), initializer=init_worker) as pool:
    for exp in EXPERIMENTS:
        if exp['name'] not in SWEEP_EXPS:
            continue
        tf, ts = load_real_targets(exp)
        chk = calibration_check(exp, tf, ts, pool)
        checks.append(chk)
        nlls = {r['mu_ratio']: r['nll'] for r in chk['rows']}
        best = min(nlls, key=nlls.get)
        print(f"  {exp['name']:>12}: NLL {nlls[0.5]:.2f}/{nlls[1.0]:.2f}/{nlls[1.5]:.2f} "
              f"-> min at mu_ratio {best} (c={chk['c']:.2f}, real_med {chk['target_s_med']:.2f})", flush=True)

# ---- 8b. SWEEP (designs x exps) ----
results = []
with _PPE(max_workers=N_WORKERS, mp_context=_mp.get_context('fork'), initializer=init_worker) as pool:
    for design in DESIGNS:
        for exp in EXPERIMENTS:
            if exp['name'] not in SWEEP_EXPS:
                continue
            tf, ts = load_real_targets(exp)
            print(f"Running {design} {exp['name']:>12} ...", end=' ', flush=True)
            r = run_experiment(exp, design, tf, ts, pool)
            results.append(r)
            print(f"mu {r['mu_true']:.2f} -> {r['mu_final']:.2f} (c={r['c']:.2f}) | "
                  f"gamma {r['gamma_true']:.1f} -> {r['gamma_final']:.2f} | "
                  f"NLL {r['nll_final']:.2f} | {r['t_elapsed'] / 60:.1f} min", flush=True)

# ---- 8c. FISHER at each fitted point ----
fisher = []
with _PPE(max_workers=N_WORKERS, mp_context=_mp.get_context('fork'), initializer=init_worker) as pool:
    for r in results:
        f = fisher_at_fitted(r, pool)
        fisher.append(f)
        print(f"FISHER {f['design']} {f['exp']:>12}: |dmu|/sig={f['dmu_sigma']:.2f} "
              f"sig_mu={f['std_mu']:.2f} | |dgamma|/sig={f['dgamma_sigma']:.2f} "
              f"sig_gamma={f['std_gamma']:.2f}", flush=True)

out = dict(mode='FULL', designs=DESIGNS, exps=SWEEP_EXPS, n_runs=N_RUNS, n_iter=N_ITER,
           m_final=M_FINAL, c_exp=C_EXP, rows=fisher)
json.dump(out, open('/tmp/19c_sweep.json', 'w'), indent=1)
json.dump(checks, open('/tmp/19c_check.json', 'w'), indent=1)
print(f'\nTotal: {(time.time() - t_all) / 60:.1f} min — saved /tmp/19c_sweep.json + /tmp/19c_check.json', flush=True)
```

```python
# ============================================================
# 4. REAL TARGETS (15-series load & filter) + machinery import
# ============================================================
def load_real_targets(exp):
    d = np.genfromtxt(exp['data_file'])
    fwhm_mhz = d[:, 0] * 1000.0
    err_mhz = d[:, 1] * 1000.0
    ok = ~np.isnan(fwhm_mhz) & ~np.isnan(err_mhz) & (fwhm_mhz > 0)
    filt = ok & (err_mhz / fwhm_mhz < 10.0)
    target_f = torch.tensor(fwhm_mhz[filt], dtype=torch.float32)
    target_s = torch.tensor(err_mhz[filt], dtype=torch.float32)
    if TARGET_CAP is not None and len(target_f) > TARGET_CAP:
        idx = np.random.default_rng(12345).choice(len(target_f), TARGET_CAP, replace=False)
        target_f = target_f[idx]
        target_s = target_s[idx]
    return target_f, target_s

def target_bandwidths(target_f, target_s):
    n = len(target_f)
    SCOTT = n ** (-1.0 / 6.0)
    H_F = float(target_f.std()) * SCOTT
    H_S = max(float(target_s.std()) * SCOTT, H_S_MIN)
    return H_F, H_S

print('real-target loader ready')
```

```python
# ============================================================
# 5. LIKELIHOOD + μ-SCORE (D1, count-calibrated) + c(T) σ-scale
# ============================================================
def kde_scores(sim_f, sim_s, sim_n, sim_df, sim_ds, data_f, data_s, h_f, h_s, mu, sigma_prop, gamma_scale=True):
    d_f = data_f[:, None] - sim_f[None, :]
    d_s = data_s[:, None] - sim_s[None, :]
    W = torch.exp(-0.5 * (d_f / h_f) ** 2 - 0.5 * (d_s / h_s) ** 2)
    w = W / W.sum(dim=1, keepdim=True).clamp_min(1e-12)
    score = (sim_n[None, :] - mu) / sigma_prop ** 2
    s_mu = (w * score).sum(dim=1)
    if gamma_scale:
        dlogG = ((d_f * sim_df[None, :]) / h_f + (d_s * sim_ds[None, :]) / h_s) / H_REF
    else:
        dlogG = (d_f * sim_df[None, :]) / h_f ** 2 + (d_s * sim_ds[None, :]) / h_s ** 2
    s_gamma = (w * dlogG).sum(dim=1)
    logp = torch.log((W.sum(dim=1) / len(sim_f)).clamp_min(1e-30))
    return s_mu, s_gamma, -logp.mean(), w

def grad_mu_design(score_variant, w, sim_n, mu, sigma_prop):
    '''D1 scalar μ gradient (σ_ref = σ_prop).'''
    B = w.mean(dim=0)
    score = (sim_n - mu) / sigma_prop ** 2
    return float(max(min(-(B - B.mean()) @ score, CLIP), -CLIP))

def mu_score_point(score_variant, w, sim_n, mu, sigma_prop):
    '''D1 per-data-point μ score (Fisher).'''
    score = (sim_n[None, :] - mu) / sigma_prop ** 2
    return (w * score).sum(dim=1)

print('likelihood + D1 score ready')
```

```python
# ============================================================
# 6. ONE EXPERIMENT: REAL targets, σ-scaled simulator (c=1 or c(T)), D1 μ score
# ============================================================
def run_experiment(exp, design, target_f, target_s, pool):
    score_variant, c = DESIGN_SPEC[design]
    if c is None:
        c = c_law(exp)
    mu_true, sigma_prop = exp['mu_true'], exp['sigma_prop']
    lam, gamma_true = exp['lam'], exp['gamma_true']
    mu_init, gamma_init = 0.5 * mu_true, 0.5 * gamma_true
    H_F, H_S = target_bandwidths(target_f, target_s)

    mu_val, gamma_val = float(mu_init), float(gamma_init)
    history = []
    t0 = time.time()
    for step in range(N_ITER):
        rng2 = np.random.default_rng(SEED + step)
        tasks, ns = [], []
        for _ in range(N_RUNS):
            u, b, n = draw_fixed_noise(mu_val, sigma_prop, lam, rng2)
            tasks.append((gamma_val, u.numpy(), b.numpy()))
            ns.append(n)
        res = parallel_map(pool, tasks)
        ft = torch.tensor([r[0] for r in res], dtype=torch.float32)
        si_t = torch.tensor([r[1] for r in res], dtype=torch.float32) * c   # σ-scale
        nt = torch.tensor(ns, dtype=torch.float32)
        dg_t = torch.tensor([r[2] for r in res], dtype=torch.float32)
        ds_t = torch.tensor([r[3] for r in res], dtype=torch.float32)

        s_mu, s_gamma, nll_val, w = kde_scores(ft, si_t, nt, dg_t, ds_t, target_f, target_s,
                                               H_F, H_S, mu_val, sigma_prop, gamma_scale=True)
        grad_mu = grad_mu_design(score_variant, w, nt, mu_val, sigma_prop)
        grad_gamma = float(max(min(-s_gamma.mean(), CLIP), -CLIP))

        lr_mu_decay = LR_MU * (1.0 - step / N_ITER)
        mu_val -= lr_mu_decay * grad_mu
        mu_val = max(1.0, min(200.0, mu_val))
        gamma_val -= LR_GAMMA * (1.0 - GAMMA_ANNEAL * step / N_ITER) * grad_gamma
        gamma_val = max(0.1, min(100.0, gamma_val))
        history.append(dict(step=step, mu=mu_val, gamma=gamma_val, nll=float(nll_val)))

    return dict(exp=exp['name'], power=exp['power'], T=exp['T'], design=design, c=c,
                score_variant=score_variant,
                mu_true=mu_true, gamma_true=gamma_true, sigma_prop=sigma_prop, lam=lam,
                mu_init=mu_init, gamma_init=gamma_init, mu_final=mu_val, gamma_final=gamma_val,
                nll_final=float(nll_val), t_elapsed=time.time() - t0,
                target_f=target_f, target_s=target_s, H_F=H_F, H_S=H_S, history=history)

print('run_experiment ready')
```

```python
# ============================================================
# 7. FISHER / CRB at the fitted point (design's own μ score, scaled σ;
#    raw γ-score for 15-series comparability) + spread diagnostic
# ============================================================
def fisher_at_fitted(r, pool):
    mu_f, gamma_f = r['mu_final'], r['gamma_final']
    sp, lam = r['sigma_prop'], r['lam']
    score_variant, c = r['score_variant'], r['c']
    rng = np.random.default_rng(FISHER_BASE_SEED)
    tasks, ns = [], []
    for _ in range(M_FINAL):
        u, b, n = draw_fixed_noise(mu_f, sp, lam, rng)
        tasks.append((gamma_f, u.numpy(), b.numpy()))
        ns.append(n)
    res = parallel_map(pool, tasks)
    ft = torch.tensor([x[0] for x in res], dtype=torch.float32)
    si_t = torch.tensor([x[1] for x in res], dtype=torch.float32) * c
    nt = torch.tensor(ns, dtype=torch.float32)
    dg_t = torch.tensor([x[2] for x in res], dtype=torch.float32)
    ds_t = torch.tensor([x[3] for x in res], dtype=torch.float32)
    target_f, target_s = r['target_f'], r['target_s']

    _, s_gamma, _, w = kde_scores(ft, si_t, nt, dg_t, ds_t, target_f, target_s,
                                  r['H_F'], r['H_S'], mu_f, sp, gamma_scale=False)
    s_mu = mu_score_point(score_variant, w, nt, mu_f, sp)
    s = torch.stack([s_mu, s_gamma], dim=1)
    J = s.T @ s / len(target_f)
    Jinv = torch.linalg.inv(J + 1e-8 * torch.eye(2))
    std_mu = float(math.sqrt(Jinv[0, 0]))
    std_gamma = float(math.sqrt(Jinv[1, 1]))
    corr = float(Jinv[0, 1] / torch.sqrt(Jinv[0, 0] * Jinv[1, 1]))
    spread_ratio = float(ft.std()) / float(target_f.std())
    return dict(exp=r['exp'], power=r['power'], T=r['T'], design=r['design'], c=c,
                mu_f=mu_f, gamma_f=gamma_f,
                mu_true=r['mu_true'], gamma_true=r['gamma_true'], n_target=len(target_f),
                std_mu=std_mu, std_gamma=std_gamma, corr=corr, spread_ratio=spread_ratio,
                dmu_sigma=abs(mu_f - r['mu_true']) / std_mu,
                dgamma_sigma=abs(gamma_f - r['gamma_true']) / std_gamma)

print('fisher_at_fitted ready')
```

```python
# ============================================================
# 8. RUN: 2 designs x 14 experiments, then Fisher
# ============================================================
import multiprocessing as _mp
from concurrent.futures import ProcessPoolExecutor as _PPE

N_CPUS = os.cpu_count() or 4
N_WORKERS = 4

t_all = time.time()
results = []
with _PPE(max_workers=N_WORKERS, mp_context=_mp.get_context('fork'), initializer=init_worker) as pool:
    for design in DESIGNS:
        for exp in EXPERIMENTS:
            if exp['name'] not in SWEEP_EXPS:
                continue
            tf, ts = load_real_targets(exp)
            print(f"Running {design} {exp['name']:>12} ...", end=' ', flush=True)
            r = run_experiment(exp, design, tf, ts, pool)
            results.append(r)
            print(f"mu {r['mu_true']:.2f} -> {r['mu_final']:.2f} (c={r['c']:.2f}) | "
                  f"gamma {r['gamma_true']:.1f} -> {r['gamma_final']:.2f} | "
                  f"NLL {r['nll_final']:.2f} | {r['t_elapsed'] / 60:.1f} min", flush=True)

# ---- Fisher at each fitted point ----
fisher = []
with _PPE(max_workers=N_WORKERS, mp_context=_mp.get_context('fork'), initializer=init_worker) as pool:
    for r in results:
        f = fisher_at_fitted(r, pool)
        fisher.append(f)
        inside = 'IN' if abs(f['mu_f'] - f['mu_true']) <= f['std_mu'] else 'OUT'
        print(f"FISHER {f['design']} {f['exp']:>12}: mu {f['mu_f']:.1f}/{f['mu_true']:.1f} "
              f"sig_mu={f['std_mu']:.1f} |dmu|/sig={f['dmu_sigma']:.2f} [{inside}] | "
              f"|dgamma|/sig={f['dgamma_sigma']:.2f}", flush=True)

hist_by = {(r['design'], r['exp']): r['history'] for r in results}
for f in fisher:
    f['history'] = hist_by.get((f['design'], f['exp']))
out = dict(mode='FULL', designs=DESIGNS, exps=SWEEP_EXPS, n_runs=N_RUNS, n_iter=N_ITER,
           m_final=M_FINAL, c_law={p: list(v) for p, v in C_LAW.items()}, rows=fisher)
os.makedirs('data/processed', exist_ok=True)
RESULT_PATH = 'data/processed/19d_sweep_history.json'
json.dump(out, open(RESULT_PATH, 'w'), indent=1)
print(f'\nTotal: {(time.time() - t_all) / 60:.1f} min — saved {RESULT_PATH}', flush=True)
```

```python
# ============================================================
# 9b. FIG 6 — (mu, gamma) phase-space optimization paths + Fisher 1sigma ellipses
#     D1 (c=1) blue, C1 (c(T)) red | green dot = start | star = end
#     dashed = truth | ellipse = Fisher at the endpoint, clips at the panel edge
#     when sigma_mu is huge (the honest-wide signature)
# ============================================================
from matplotlib.patches import Ellipse as _Ellipse, Patch as _Patch
from matplotlib.lines import Line2D as _Line2D

def fisher_ellipse_artist(r, n_sigma=1.0, **kw):
    sx, sy, rho = r['std_mu'], r['std_gamma'], r['corr']
    tr = sx * sx + sy * sy
    det = max(sx * sx * sy * sy * (1.0 - rho * rho), 1e-30)
    disc = max(tr * tr - 4.0 * det, 0.0)
    lam1 = 0.5 * (tr + math.sqrt(disc))
    lam2 = 0.5 * (tr - math.sqrt(disc))
    theta = 0.5 * math.atan2(2.0 * rho * sx * sy, sx * sx - sy * sy)
    return _Ellipse((r['mu_f'], r['gamma_f']),
                    width=2.0 * n_sigma * math.sqrt(lam1),
                    height=2.0 * n_sigma * math.sqrt(lam2),
                    angle=math.degrees(theta), **kw)

def row_for(design, exp):
    return next((r for r in sweep['rows'] if r['design'] == design and r['exp'] == exp), None)

n_exps = len(EXPS)
fig, axes = plt.subplots(4, 4, figsize=(18, 16))
for j, e in enumerate(EXPS):
    ax = axes[j // 4][j % 4]
    mus_all, gams_all = [], []
    for d in ['D1', 'C1']:
        r = row_for(d, e)
        if r is None or not r.get('history'):
            continue
        mus = [h['mu'] for h in r['history']]
        gams = [h['gamma'] for h in r['history']]
        mus_all += mus
        gams_all += gams
        ax.plot(mus, gams, '-', color=DCOL[d], lw=1.3, alpha=0.85, zorder=3)
        ax.plot(mus[::10], gams[::10], '.', color=DCOL[d], ms=2.5, alpha=0.6, zorder=3)
        ax.add_patch(fisher_ellipse_artist(r, facecolor=DCOL[d], alpha=0.22,
                                           edgecolor=DCOL[d], lw=1.2, zorder=2))
        ax.plot(mus[-1], gams[-1], '*', ms=14, color=DCOL[d], mec='k', mew=0.6, zorder=5)
        ax.text(0.98, 0.965 if d == 'D1' else 0.885,
                f"{d} |Δμ|/σ={r['dmu_sigma']:.2f}  σ_μ={r['std_mu']:.1f}",
                transform=ax.transAxes, ha='right', va='top', fontsize=7.5, color=DCOL[d])
    r0 = row_for('D1', e) or row_for('C1', e)
    if r0 is not None:
        ax.axvline(r0['mu_true'], color='g', ls='--', lw=1, alpha=0.5)
        ax.axhline(r0['gamma_true'], color='g', ls='--', lw=1, alpha=0.5)
        mus_all.append(r0['mu_true'])
        gams_all.append(r0['gamma_true'])
        if r0.get('history'):
            h0 = r0['history'][0]
            ax.plot(h0['mu'], h0['gamma'], 'o', ms=8, color='#2ca02c', mec='k', mew=0.6, zorder=6)
    if mus_all:
        m0, m1 = min(mus_all), max(mus_all)
        sp = (m1 - m0) or 1.0
        ax.set_xlim(m0 - 0.08 * sp, m1 + 0.08 * sp)
        g0, g1 = min(gams_all), max(gams_all)
        spg = (g1 - g0) or 1.0
        ax.set_ylim(g0 - 0.08 * spg, g1 + 0.08 * spg)
    ax.set_title(e, fontsize=9)
    ax.set_xlabel('μ')
    ax.set_ylabel('γ (MHz)')
    ax.grid(alpha=0.25)
for j in range(n_exps, 16):
    axes[j // 4][j % 4].axis('off')

handles = [_Line2D([], [], color=DCOL['D1'], lw=1.4, label='D1 path (σ-scale c=1)'),
           _Line2D([], [], color=DCOL['C1'], lw=1.4, label='C1 path (c = c(T) law)'),
           _Line2D([], [], marker='o', color='w', mfc='#2ca02c', mec='k', ms=7, label='start (½·truth init)'),
           _Line2D([], [], marker='*', color='w', mfc='k', mec='k', ms=11, label='end (μ̂, γ̂)'),
           _Line2D([], [], color='g', ls='--', lw=1, label='truth (μ_true, γ_true)'),
           _Patch(facecolor='#888888', alpha=0.25, label='Fisher 1σ ellipse @ endpoint')]
fig.legend(handles=handles, loc='lower center', ncol=3, fontsize=9, frameon=False,
           bbox_to_anchor=(0.5, 0.012))
fig.suptitle('19d — optimization paths in (μ, γ) phase space + Fisher 1σ ellipses '
             '(honest-wide: ellipse clips at the panel edge when σ_μ ≫ path scale)', fontsize=12)
fig.tight_layout(rect=[0, 0.045, 1, 0.955])
plt.show()
```

### Failure mode #8 — final write-up (un-damped σ-channel gradient)

**Definition:** tight-wrong point estimates produced by score designs whose only *live* gradient comes from a systematically mismatched channel.

**Evidence chain:**
- 19b D2 (raw σ-gradient, σ_ref=10): collapses μ to the floor — |Δμ|/σ up to **557** (3nW T100).
